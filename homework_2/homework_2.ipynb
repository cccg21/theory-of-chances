### Задание 1.38
> Пятитомное собрание сочинений расположено на полке в
случайном порядке. Какова вероятность того, что книги стоят слева
направо в порядке нумерации томов (от 1 до 5)?
* Рассмотрим событие A - книги стоят слева направо в порядке нумерации томов (от 1 до 5).

$P(A) = \frac{n}{m} = \frac{1}{A_{5}^5} = \frac{1}{5!}$

```python
result = round(1 / sc.factorial(5), 4)
print(result)
```

### Задание 1.43
> Наудачу взятый телефонный номер состоит из 5 цифр.
Какова вероятность того, что в нем все цифры: а) различные;
б) одинаковые; в) нечетные? Известно, что номер телефона не начинается с цифры ноль.
* Рассмотрим событие A - цифры различные.

$P(A) = \frac{n}{m} = \frac{9 \cdot {A_{9}^4}}{9 \cdot 10^4}$

```python
r_A = round( (9 * sc.perm(9,4)) / (9 * 10**4), 4)
print(r_A)
```

* Рассмотрим событие B - цифры одинаковые.

$P(B) = \frac{n}{m} = \frac{{C_{9}^1}}{9 \cdot 10^4}$

```python
r_B = round(sc.comb(9, 1) / (9 * 10**4), 4)
print(r_B)
```

* Рассмотрим событие C - цифры нечетные.

$P(C) = \frac{n}{m} = \frac{\overline{A_{5}^5}}{9 \cdot 10^4}$

```python
r_C = round(5**5 / (9 * 10**4), 4)
print(r_C)
```

### Задание 1.48
> В старинной игре в кости необходимо было для выигрыша
получить при бросании трех игральных костей сумму очков, превосходящую 10. Найти вероятности: а) выпадения 11 очков; б) выигрыша.
* Рассмотрим событие A - выпадение 11 очков.

```python
import itertools
```

```python
arr = list(itertools.product(range(1, 7), repeat = 3))

summa_11 = [x for x in arr if sum(x) == 11]
r_A = round(len(summa_11) / len(arr), 4)
print(r_A)
```

* Рассмотрим событие B - выигрыш.

```python
arr = list(itertools.product(range(1,7), repeat = 3))

n = [x for x in arr if sum(x) > 10]
r_B = round(len(n) / len(arr), 4)
print(r_B)
```
