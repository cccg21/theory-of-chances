### Тема 1. Моделирование геометрической вероятности
> Смоделируйте вычисление геометрической вероятности (на примере из лекции про метание дротиков в игровой круг, площадь которого в два раза меньше площади стены). Показать теоретический расчет и графическую иллюстрацию для рассматриваемого примера.


### Задача.
>Палку длиной 1 наугад ломают в двух случайных точках (независимо и равномерно на отрезке [0, 1]). Получаются три куска. Какова вероятность того, что из этих трёх кусков можно сложить треугольник?

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle, Polygon
import numpy as np

def broken_stick(n):
    a = 1 

    fig, ax = plt.subplots(figsize=(5, 5))
    ax = plt.gca()

    ax.add_patch(Rectangle((0, 0), a, a, facecolor='g'))

    tri1 = Polygon([[0, a/2], [a/2, a/2], [a/2, a]]) # x < y
    tri2 = Polygon([[a/2, 0], [a/2, a/2], [a, a/2]]) # x > y 
    ax.add_patch(tri1)
    ax.add_patch(tri2)


    X = np.random.uniform(0, a, n)
    Y = np.random.uniform(0, a, n)
    plt.scatter(X, Y, marker='.', c="black")

    inside = []
    for x, y in zip(X, Y):
        low, high = min(x, y), max(x, y)
        p1, p2, p3 = low, high - low, a - high
        if p1 < a/2 and p2 < a/2 and p3 < a/2:
            inside.append([x, y])

    prob = len(inside) / n

    plt.xlabel("точка разлома x")
    plt.ylabel("точка разлома y")
    plt.show()

    print(f"Из {n} точек в область 'треугольник существует' попало {len(inside)}. "
          f"Доля попавших (оценка P) = {prob}")
    print(f"Теоретическая вероятность = {1/4}")

broken_stick(1000)
```
