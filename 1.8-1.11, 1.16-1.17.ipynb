### Задача 1.8
> В конкурсе по 5 номинациям участвуют 10 кино - фильмов. Сколько существует вариантов распределения призов, если по каждой номинации установлены: а) различные призы; б) одинаковые призы?
 * Рассмотрим событие А - различные призы.

$P(A) = \overline {A_{n}^k} = n^k = 10^5$

```python
A = 10 ** 5
print('A =', A)
```

 * Рассмотрим событие B - одинаковые призы.
$P(B) = \overline{C_{10}^5} = {C_{10+5-1}^5} = {C_{14}^5} = \frac{14!}{5!(14-5)!} = \frac{14!}{5! \cdot 9!}$

```python
B = sc.comb(10, 5, repetition=True)
print('B =', B)
```

### Задача 1.9
> Сколько существует семизначных чисел, состоящих из цифр 4, 5 и 6, в которых цифра 4 повторяется 3 раза, а цифры 5 и 6 - по 2 раза?
* Рассмотрим событие ${P_{7}(3; 2; 2)} = \frac{7!}{3! \cdot 2! \cdot 2!} $

```python
P = sc.factorial(7) / ( sc.factorial(3) * sc.factorial(2) * sc.factorial(2) )
print('P =', P)
```

### Задача 1.10
> Буквы Т, Е, И, Я, Р, О написаны на отдельных карточках. Ребенок берет карточки в случайном порядке и прикладывает одну к друтой: а) 3 карточки; б) все 6 карточек. Какова вероятность того, что получится слово: а) «ТОР»; б) «ТЕОРИЯ»?
* Рассмотрим событие A - «ТОР» : $P(A) = \frac{n}{m} = \frac{1}{{A_{6}^3}} = \frac{1}{6 \cdot 5 \cdot 4}$

```python
P = round(1 / sc.perm(6, 3), 5
         )
print('P =', P)
```

* Рассмотрим событие B - «ТЕОРИЯ» : $P(B) = \frac{n}{m} = \frac{1}{{P_{6}}} = \frac{1}{6!}$

```python
P = round(1 / sc.perm(6, 6), 5)
print('P =', P)
```

### Задача 1.11
> Используя условие примера 1.10, найти вероятность того, что получится слово «АНАНАС», если на отдельных карточках написаны три буквы А, две буквы Н и одна буква С.

### Задача 1.16
> За круглым столом рассаживаются 5 мужчин и 5 женщин. Найти вероятность того, что: а) никакие два лица одного пола не сядут рядом; б) супруги сядут рядом, если эти мужчины и женщины образуют 5 супружеских пар.
* Рассмотрим событие A - никакие два лица одного пола не сядут рядом :
$P(A) = \frac{n}{m} = \frac{2 \cdot (5!)^2}{{P_{10}}}$

```python
P = round(( 2 * sc.factorial(5) ** 2) / ( sc.perm(10, 10)) , 5)
print(P)
```

* Рассмотрим событие B - супруги сядут рядом, если эти мужчины и женщины образуют 5 супружеских пар.
$P(B) = \frac{5! \cdot 2^5}{{P_{10}}}$

```python
P = round(( sc.factorial(5) * 2**5) / ( sc.perm(10, 10) ), 5)
print(P)
```

### Задача 1.17
> В купейный вагон (9 купе по 4 места) семи пассажирам продано 7 билетов. Найти вероятности того, что пассажиры попали: а) в два купе; б) в семь купе; в) в три купе.
* Рассмотрим событие A - пассажиры попали в 2 купе.
$P(A) = \frac{{C_{9}^2} \cdot {C_{8}^7}}{{C_{36}^7}}$

```python
P = round(( sc.comb(9, 2) * sc.comb(8, 7) ) / sc.comb(36, 7), 7)
print('P =', P)
```

* Рассмотрим событие B - пассажары попали в 7 купе.
$P(A) = \frac{{C_{9}^7} \cdot 4^7}{{C_{36}^7}}$

```python
P = round((sc.comb(9,7) * (4**7) / sc.comb(36,7)), 7)
print('P =', P)
```

* Рассмотрим событие C - пассажиры попали в 3 купе. Действительно, возможные варианты выбора 7 мест из 12 в трех купе следующие: 4 + 2 + 1, 3 + 3 + 1, 3 + 2 + 2, а за счет перестановок купе таких вариантов будет соответственно 6, 3 и 3. Каждый из этих вариантов по правилу произведения может быть получен $С_{4}^{4}{{C}}_{4}^{2}{{C}}_{4}^{1},\ С_{4}^{3}{{C}}_{4}^{3}{{C}}_{4}^{1},\ С_{4}^{3}{{C}}_{4}^{2}{{C}}_{4}^{2}$ способами соответственно.
$${P}({C})\ =\frac{{{C}}_{9}^{3}(6{{C}}_{4}^{4}{{C}}_{4}^{2}{{C}}_{4}^{1}+3{{C}}_{4}^{3}{{C}}_{4}^{3}{{C}}_{4}^{1}+3{{C}}_{4}^{3}{{C}}_{4}^{2}{{C}}_{4}^{2})}{{{C}}_{36}^{7}}$$$$$$

```python
P = round(sc.comb(9, 3) * (6 * sc.comb(4, 4) * sc.comb(4, 2) * sc.comb(4, 1) +
                 3 * sc.comb(4, 3) * sc.comb(4, 3) * sc.comb(4, 1) +
                 3 * sc.comb(4, 3) * sc.comb(4, 2) * sc.comb(4, 2)) / sc.comb(36, 7), 7)
print('P =', P)
```
